```python
#국 메뉴
import random
menu_soup = ['청국장', '동태찌게', '쇠고기무국', '만두국', '우거지해장국', '추어탕', '뼈해장국', '버섯불고기', '육계장', '해물순두부', '삼계탕', '김치찌게', '도가니설렁탕',
             '갈비탕', '장어', '오리고기']
print(menu_soup)
```

```python
#국 메뉴 중 한가지를 선택
print(random.choice(menu_soup))
```

```python
#점심메뉴 중 한가지를 선택
print(random.choice(menu_lunch))
```

```python
#1~18 사이의 숫자 맞추기
import random
randomNum = random.randint(1, 18)

guess = int(input('1과 18사이의 숫자를 하나 입력하세요: '))
count = 1

while guess != randomNum :
    if guess > randomNum :
        print('더 작은 수를 선택하세요')
    else :
        print('더 큰 수를 선택하세요')

    guess = int(input('숫자를 맞춰보세요(1-18): '))
    count += 1

print('성공하셨군요', str(count)+'번 만에 맞췄습니다')
print('입력한 숫자는', str(randomNum)+'입니다')
```
